# 1. Load Raw Dataset

The primary dataset contains LinkedIn job postings along with salary, company, location, and job description information.

```python
postings = pd.read_csv("../data/raw/linkedin_job_postings/postings.csv")

postings.head()
```

```python
print("Shape:", postings.shape)

postings.info()
```

```python
postings.describe(include="all")
```

# 2. Missing Value Analysis

Understanding missing values is essential before data cleaning.

```python
missing = (
    postings.isnull()
    .sum()
    .sort_values(ascending=False)
)

missing_percentage = (
    postings.isnull().mean() * 100
).sort_values(ascending=False)

pd.DataFrame({
    "Missing Count": missing,
    "Missing Percentage": missing_percentage
}).head(15)
```

```python
missing_percentage.head(15).plot(
    kind="bar",
    figsize=(12,5)
)

plt.title("Top Missing Value Columns")

plt.ylabel("Percentage")

plt.show()
```
